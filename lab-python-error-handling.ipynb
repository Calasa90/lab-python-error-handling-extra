The try-except block in Python is designed to handle exceptions and provide a fallback mechanism when code encounters errors. By enclosing the code that could potentially throw errors in a try block, followed by specific or general exception handling in the except block, we can gracefully recover from errors and continue program execution.

However, there may be cases where an input may not produce an immediate error, but still needs to be addressed. In such situations, it can be useful to explicitly raise an error using the "raise" keyword, either to draw attention to the issue or handle it elsewhere in the program.

Modify the code to handle possible errors in Python, it is recommended to use `try-except-else-finally` blocks, incorporate the `raise` keyword where necessary, and print meaningful error messages to alert users of any issues that may occur during program execution.



```python
def get_unique_list(lst):

    # TRY = intentamos ejecutar el código que queremos controlar.
    try:

        # isinstance(lst, list) comprueba si "lst" es una lista.
        # Añadimos "not" porque queremos detectar el caso contrario:
        # si lst NO es una lista.
        if not isinstance(lst, list):

            # Python no considera necesariamente un error que le pasemos
            # otro tipo de dato, así que provocamos nosotros el error
            # porque nuestra función espera específicamente una lista.
            raise ValueError("The argument must be a list.")


    # EXCEPT = se ejecuta si dentro del try ocurre un ValueError.
    except ValueError as error:

        # Mostramos el mensaje del error que hemos provocado con raise.
        print(f"Error: {error}")


    # ELSE = se ejecuta solamente si el try termina SIN errores.
    else:

        # set(lst) elimina los elementos duplicados.
        # Después list() vuelve a convertir el resultado en una lista.
        return list(set(lst))


    # FINALLY = se ejecuta siempre:
    # tanto si hubo un error como si todo funcionó correctamente.
    finally:

        # En este ejercicio lo usamos para practicar finally.
        # En un programa real podría utilizarse, por ejemplo,
        # para cerrar un archivo o liberar algún recurso.
        print("Finished")
```

```python
result = get_unique_list([1, 2, 2, 3])
print(result)
```

```python
result = get_unique_list("BTS")
print(result)
```

```python
def count_case(text):

    # TRY = intentamos ejecutar el código que queremos controlar.
    try:

        # isinstance(text, str) comprueba si "text" es un string.
        # Usamos "not" porque queremos detectar el caso incorrecto:
        # que text NO sea un string.
        if not isinstance(text, str):

            # Como el problema es que hemos recibido un TIPO de dato
            # incorrecto, provocamos un TypeError.
            raise TypeError("The argument must be a string.")


    # EXCEPT = se ejecuta si dentro del try ocurre un TypeError.
    except TypeError as error:

        # Mostramos el mensaje del error.
        print(f"Error: {error}")


    # ELSE = se ejecuta solamente si el try termina SIN errores.
    else:

        # Creamos dos contadores que empiezan en 0.
        # Uno contará las mayúsculas y otro las minúsculas.
        upper = 0
        lower = 0

        # Recorremos el texto carácter por carácter.
        for character in text:

            # Si el carácter es una letra mayúscula,
            # aumentamos el contador de mayúsculas en 1.
            if character.isupper():
                upper += 1

            # Si no era mayúscula, comprobamos si es minúscula.
            # Si lo es, aumentamos el contador de minúsculas en 1.
            elif character.islower():
                lower += 1

        # Devolvemos los dos resultados.
        # Python los devuelve juntos como una tupla: (upper, lower).
        return upper, lower


    # FINALLY = se ejecuta SIEMPRE,
    # tanto si hubo un error como si todo funcionó correctamente.
    finally:

        # En este ejercicio lo usamos para practicar finally.
        print("Finished")
```

```python
result = count_case("Hello BTS")
print(result)
```

```python
result = count_case(123)
print(result)
```
